```python
import sys
sys.path.append('../src')

# Count damage distribution in training vs validation
from dataset import DisasterDataset, get_transforms
from torch.utils.data import random_split

train_transform, _ = get_transforms()
full_dataset = DisasterDataset('../data/train', transform=train_transform)

# Same split as training
val_size = int(0.2 * len(full_dataset))
train_size = len(full_dataset) - val_size
train_dataset, val_dataset = random_split(full_dataset, [train_size, val_size])

# Count in training set
train_damage_counts = {0: 0, 1: 0, 2: 0, 3: 0}
for i in range(len(train_dataset)):
    _, _, label = train_dataset[i]
    train_damage_counts[label] += 1

print("Training set distribution:")
for level, count in train_damage_counts.items():
    labels = ['no-damage', 'minor', 'major', 'destroyed']
    pct = (count / len(train_dataset)) * 100
    print(f"  {labels[level]}: {count} ({pct:.1f}%)")

# Count in validation set
val_damage_counts = {0: 0, 1: 0, 2: 0, 3: 0}
for i in range(len(val_dataset)):
    _, _, label = val_dataset[i]
    val_damage_counts[label] += 1

print("\nValidation set distribution:")
for level, count in val_damage_counts.items():
    pct = (count / len(val_dataset)) * 100
    print(f"  {labels[level]}: {count} ({pct:.1f}%)")
```
